```python
S0 = 1.10
K = 1.05
r = 0.03
T = 1
kappa = 2
lambdA = 0
theta = 0.01
sig_0 = 0.10
sig_sig = 0.2
rho = -0.1
ks = kappa + lambdA
ts = kappa*theta / ks

N=256
L =12

X = np.log(S0/K)

print("kappa_star =", ks)
print("theta_star =", ts)
print("x =", X)
```

```python
def trunc_range(r,T,kappa,lambd,theta,sig_0,sig_sig,rho,L,x):
    """
    Calcule les cumulants  puis les bornes de l'intervalle de troncature.
    """

    kappa_star=kappa+lambd
    theta_star=kappa*theta/(kappa+lambd)

    c_1=r*T+ (1-np.exp(-kappa_star*T))*(theta_star-sig_0**2)/(2*kappa_star) -(1/2)*theta_star*T
    # par souci de lisibilité on sépare en 5 le calcul de c_2.
    A=sig_sig*T*kappa_star*np.exp(-kappa_star*T)*(sig_0**2-theta_star)*(8*kappa_star*rho-4*sig_sig)
    B=kappa_star*rho*sig_sig*(1-np.exp(-kappa_star*T))*(16*theta_star-8*sig_0**2)
    C=2*theta_star*kappa_star*T*(-4*kappa_star*rho*sig_sig+sig_sig**2+4*kappa_star**2)
    D=sig_sig**2*((theta_star-2*sig_0**2)*np.exp(-2*kappa_star*T)+theta_star*(6*np.exp(-kappa_star*T)-7)+2*sig_0**2)
    E=8*kappa_star**2*(sig_0**2-theta_star)*(1-np.exp(-kappa_star*T))
    c_2=(1/(8*kappa_star**3))*(A+B+C+D+E)

    a=x+c_1-L*np.sqrt(c_2)
    b=x+c_1+L*np.sqrt(c_2)
    return a, b, c_1, c_2
```

```python
a, b, c_1, c_2 = trunc_range(
    r, T, kappa, lambdA, theta, sig_0, sig_sig, rho, L, X
)
print("r =", r, "T =", T)
print("kappa =", kappa, "lambd =", lambdA, "theta =", theta)
print("sig_0 =", sig_0, "sig_sig =", sig_sig, "rho =", rho)
print("L =", L, "x =", X)
print("c_1 =", c_1)
print("c_2 =", c_2)
print("a =", a)
print("b =", b)
```

```python

k = np.arange(N)
omega = k*np.pi/(b-a)

print("Nombre de fréquences :", len(omega))
print("Premiers indices :", k[:5])
print("Premières fréquences :", omega[:5])
```

```python
def carac_heston (omega, r, T, ks, ts, sig_0, sig_sig, rho):

  h = ks- 1j * rho * sig_sig * omega
  D = np.sqrt(h**2 + (omega**2+ 1j * omega)* sig_sig**2)
  G = (h-D)/(h+D)
  Exp1 = 1j * omega * r * T + ((sig_0**2)/(sig_sig**2))*((1-np.exp(-D*T))/(1-G*np.exp(-D*T))) * (h-D)

  Exp2 = (ks*ts/sig_sig**2) * (T*(h-D) - 2*np.log((1-G*np.exp(-D*T))/(1-G)))

  return np.exp(Exp1 + Exp2)
```

```python
phi = carac_heston(
    omega, r, T, ks, ts, sig_0, sig_sig, rho
)

print("Nombre de valeurs :", len(phi))
print("Premières valeurs :", phi[:5])
print("Valeur en omega = 0 :", phi[0])
```

```python
def chi (omega, a,p,q) :
  chi_q = np.exp(q)*(np.cos(omega*(q-a)) +omega * np.sin(omega*(q-a)))
  chi_p = np.exp(p)*(np.cos(omega*(p-a)) +omega * np.sin(omega*(p-a)))

  return (chi_q-chi_p)/(1+omega**2)


def psi (omega,a,p,q) :
  res = np.empty_like(omega)

  res[0] = q-p
  w = omega[1:]
  res[1:] = (np.sin(w*(q-a)) - np.sin(w*(p-a)))/w

  return res
```

```python
chi_put = chi(omega, a, a, 0.0)
psi_put = psi(omega, a, a, 0.0)

U_put = 2 / (b - a) * (psi_put - chi_put)
```

```python
print("Test chi :", np.isclose(chi_put[0], 1 - np.exp(a)))
print("Test psi :", np.isclose(psi_put[0], -a))
print("Premiers U_put :", U_put[:5])
```

```python
def P(omega, phi, U_put, x, a, K, r, T):
  fact_e = np.exp(1j*omega*(x-a))

  termes = phi*U_put*fact_e
  poids = np.ones_like(omega)
  poids[0] = 0.5
  somme = np.sum(poids * termes)

  return K*np.exp(-r*T)*np.real(somme)
```

```python
prix_put = P(omega, phi, U_put, X, a, K, r, T)
print("Prix du put :", prix_put)
```

```python
# Vérification de l'étape 6 avec les fonctions déjà définies :
# carac_heston, chi, psi et P

assert S0 > 0 and K > 0, "S0 et K doivent être positifs."
assert T > 0, "Ce test suppose T > 0."
assert a < 0 < b, "Les coefficients utilisés supposent a < 0 < b."
assert np.isclose(X, np.log(S0 / K)), "x ne correspond pas à log(S0/K)."

# Bornes théoriques du prix d'un put européen sans dividendes
borne_inf = max(K * np.exp(-r*T) - S0, 0.0)
borne_sup = K * np.exp(-r*T)

print(f"Bornes théoriques : [{borne_inf:.8f}, {borne_sup:.8f}]")
print()
print(f"{'N':>6} | {'Prix du put':>18} | {'Écart précédent':>18} | Bornes")
print("-" * 76)

prix_precedent = None
resultats = []

for n_test in [10, 32, 64, 128, 256, 512, 1024]:

    # 1. Recalculer toute la grille pour ce nombre de termes
    k_test = np.arange(n_test)
    omega_test = k_test * np.pi / (b - a)

    # 2. Recalculer la fonction caractéristique sur cette grille
    phi_test = carac_heston(
        omega_test, r, T,
        ks, ts,
        sig_0, sig_sig, rho
    )

    # 3. Recalculer les coefficients du payoff
    chi_test = chi(omega_test, a, a, 0.0)
    psi_test = psi(omega_test, a, a, 0.0)
    U_test = 2 / (b - a) * (psi_test - chi_test)

    # 4. Vérifier les tableaux
    assert phi_test.shape == omega_test.shape == U_test.shape
    assert np.all(np.isfinite(phi_test)), "Valeur invalide dans phi."
    assert np.all(np.isfinite(U_test)), "Valeur invalide dans U_put."
    assert np.isclose(phi_test[0], 1.0), "Phi(0) doit valoir 1."

    # 5. Calculer le prix avec ta fonction P
    prix_test = P(
        omega_test, phi_test, U_test,
        X, a, K, r, T
    )

    assert np.isfinite(prix_test), "Le prix n'est pas fini."

    # 6. Vérifier l'assemblage avec une autre écriture de la somme
    termes_test = (
        phi_test
        * U_test
        * np.exp(1j * omega_test * (X - a))
    )

    # Somme de tous les termes, puis retrait de la moitié du premier
    somme_reference = np.sum(termes_test) - 0.5 * termes_test[0]

    prix_reference = (
        K * np.exp(-r*T) * np.real(somme_reference)
    )

    assert np.isclose(
        prix_test, prix_reference,
        rtol=1e-12, atol=1e-10
    ), "La fonction P ne reproduit pas la somme COS attendue."

    # 7. Contrôler les bornes, avec une petite tolérance numérique
    tolerance = 1e-8 * max(1.0, S0, K)
    bornes_ok = (
        borne_inf - tolerance
        <= prix_test
        <= borne_sup + tolerance
    )

    # 8. Comparer au prix précédent
    if prix_precedent is None:
        ecart_texte = "—"
    else:
        ecart = abs(prix_test - prix_precedent)
        ecart_texte = f"{ecart:.10e}"

    print(
        f"{n_test:6d} | {prix_test:18.10f} | "
        f"{ecart_texte:>18} | {'OK' if bornes_ok else 'HORS BORNES'}"
    )

    resultats.append((n_test, prix_test))
    prix_precedent = prix_test

print()
print("Contrôles d'assemblage terminés.")
print("Écart entre N=512 et N=1024 :",
      abs(resultats[-1][1] - resultats[-2][1]))
```

```python
from scipy.stats import norm

def black_scholes(S0, K, r, T, sigma):
    """
    Renvoie (prix_put, prix_call) sans dividendes.
    """
    if S0 <= 0 or K <= 0:
        raise ValueError("S0 et K doivent être positifs.")

    if T < 0 or sigma < 0:
        raise ValueError("T et sigma doivent être positifs ou nuls.")

    # À l'échéance
    if T == 0:
        return max(K - S0, 0.0), max(S0 - K, 0.0)

    # Cas sans volatilité
    if sigma == 0:
        strike_actualise = K * np.exp(-r*T)
        return (
            max(strike_actualise - S0, 0.0),
            max(S0 - strike_actualise, 0.0)
        )

    d1 = (
        np.log(S0 / K) + (r + 0.5*sigma**2)*T
    ) / (sigma*np.sqrt(T))

    d2 = d1 - sigma*np.sqrt(T)

    call_bs = (
        S0*norm.cdf(d1)
        - K*np.exp(-r*T)*norm.cdf(d2)
    )

    put_bs = (
        K*np.exp(-r*T)*norm.cdf(-d2)
        - S0*norm.cdf(-d1)
    )

    return put_bs, call_bs
```

```python
# Choix de comparaison : volatilité constante égale à la volatilité initiale
sigma_bs = sig_0

put_bs, call_bs = black_scholes(
    S0, K, r, T, sigma_bs
)

# Recalculer les prix Heston avec tes tableaux actuels
put_heston = P(
    omega, phi, U_put, np.log(S0/K), a, K, r, T
)

call_heston = call_depuis_put(
    put_heston, S0, K, r, T
)

print("Volatilité constante Black-Scholes :", sigma_bs)
print()
print(f"{'Modèle':<18} | {'Put':>16} | {'Call':>16}")
print("-" * 58)
print(f"{'Heston-COS':<18} | {put_heston:16.10f} | {call_heston:16.10f}")
print(f"{'Black-Scholes':<18} | {put_bs:16.10f} | {call_bs:16.10f}")
print("-" * 58)
print(
    f"{'Écart Heston - BS':<18} | "
    f"{put_heston-put_bs:16.10f} | "
    f"{call_heston-call_bs:16.10f}"
)
```

```python
from scipy.optimize import brentq

def volatilite_implicite_put(prix_cible, S0, K, r, T):
    """
    Cherche la volatilité Black-Scholes qui reproduit prix_cible.
    """
    if T <= 0:
        raise ValueError("La maturité doit être strictement positive.")

    borne_inf = max(K*np.exp(-r*T) - S0, 0.0)
    borne_sup = K*np.exp(-r*T)

    # Une volatilité implicite finie et strictement positive
    # nécessite un prix strictement entre ces bornes.
    if not borne_inf < prix_cible < borne_sup:
        raise ValueError(
            f"Prix incompatible avec l'inversion pour K={K:.4f} : "
            f"{prix_cible:.10f}, bornes [{borne_inf:.10f}, "
            f"{borne_sup:.10f}]."
        )

    def ecart(sigma):
        put_bs, _ = black_scholes(S0, K, r, T, sigma)
        return put_bs - prix_cible

    sigma_min = 1e-8
    sigma_max = 1.0

    # Élargir l'intervalle de recherche si nécessaire.
    while ecart(sigma_max) < 0 and sigma_max < 16:
        sigma_max *= 2

    if ecart(sigma_min) > 0 or ecart(sigma_max) < 0:
        raise ValueError(
            f"Impossible d'encadrer la volatilité implicite pour K={K:.4f}."
        )

    return brentq(
        ecart,
        sigma_min,
        sigma_max,
        xtol=1e-12
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import norm
from scipy.optimize import brentq


def put_black_scholes(S0, K, r, T, sigma):
    """Prix d'un put européen sans dividendes sous Black-Scholes."""
    if sigma == 0:
        return max(K*np.exp(-r*T) - S0, 0.0)

    d1 = (
        np.log(S0/K) + (r + 0.5*sigma**2)*T
    ) / (sigma*np.sqrt(T))

    d2 = d1 - sigma*np.sqrt(T)

    return (
        K*np.exp(-r*T)*norm.cdf(-d2)
        - S0*norm.cdf(-d1)
    )
```

```python
def volatilite_implicite_put(prix_put, S0, K, r, T):
    """Volatilité Black-Scholes correspondant au prix du put fourni."""
    if S0 <= 0 or K <= 0 or T <= 0:
        raise ValueError("Il faut S0 > 0, K > 0 et T > 0.")

    borne_inf = max(K*np.exp(-r*T) - S0, 0.0)
    borne_sup = K*np.exp(-r*T)

    if not np.isfinite(prix_put):
        raise ValueError("Le prix du put n'est pas fini.")

    if prix_put < borne_inf or prix_put >= borne_sup:
        raise ValueError(
            f"Prix incompatible avec une volatilité implicite finie : "
            f"prix={prix_put}, bornes=[{borne_inf}, {borne_sup}["
        )

    if prix_put == borne_inf:
        return 0.0

    def objectif(sigma):
        return put_black_scholes(S0, K, r, T, sigma) - prix_put

    # Élargir la borne supérieure si nécessaire.
    sigma_max = 1.0

    while objectif(sigma_max) < 0 and sigma_max < 16.0:
        sigma_max *= 2

    if objectif(sigma_max) < 0:
        raise ValueError("Impossible d'encadrer la volatilité implicite.")

    return brentq(objectif, 0.0, sigma_max, xtol=1e-12)
```

```python
def put_heston_pour_strike(
    S0, strike, r, T,
    kappa, lambd, theta, sig_0, sig_sig, rho,
    N=512, L=12
):
    """Prix COS du put pour un strike donné."""
    ks = kappa + lambd
    ts = kappa*theta / ks

    x_local = np.log(S0 / strike)

    # Fonctionne si trunc_range renvoie (a, b)
    # ou temporairement (a, b, c_1, c_2).
    bornes = trunc_range(
        r, T, kappa, lambd, theta,
        sig_0, sig_sig, rho, L, x_local
    )

    a_local, b_local = bornes[:2]

    if not a_local < 0 < b_local:
        raise ValueError(
            "Les coefficients actuels supposent a < 0 < b."
        )

    k_local = np.arange(N)
    omega_local = k_local*np.pi / (b_local - a_local)

    phi_local = carac_heston(
        omega_local, r, T, ks, ts, sig_0, sig_sig, rho
    )

    chi_local = chi(omega_local, a_local, a_local, 0.0)
    psi_local = psi(omega_local, a_local, a_local, 0.0)

    U_local = (
        2 / (b_local - a_local)
        * (psi_local - chi_local)
    )

    return P(
        omega_local, phi_local, U_local,
        x_local, a_local, strike, r, T
    )
```

```python
# Grille de strikes : de 60 % à 140 % du spot
strikes = np.linspace(0.6*S0, 1.4*S0, 61)

prix_heston = []
vols_implicites = []

for strike in strikes:
    prix = put_heston_pour_strike(
        S0, strike, r, T,
        kappa, lambdA, theta, sig_0, sig_sig, rho,
        N=512, L=L
    )

    prix_heston.append(prix)

    try:
        iv = volatilite_implicite_put(
            prix, S0, strike, r, T
        )
    except ValueError as erreur:
        print(f"Strike {strike:.2f} : {erreur}")
        iv = np.nan

    vols_implicites.append(iv)

prix_heston = np.asarray(prix_heston)
vols_implicites = np.asarray(vols_implicites)

# Tracé en pourcentage
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    strikes,
    100*vols_implicites,
    marker="o",
    markersize=3,
    label="Volatilité implicite des prix Heston"
)

ax.axvline(
    S0,
    color="grey",
    linestyle="--",
    label=f"Spot S0 = {S0:g}"
)

ax.set_xlabel("Strike K")
ax.set_ylabel("Volatilité implicite annuelle (%)")
ax.set_title(
    f"Smile / skew de Heston — T = {T:g} ans, rho = {rho:g}"
)

ax.grid(alpha=0.3)
ax.legend()

plt.show()
```

```python
# Vérification de la parité call-put et des bornes
# Hypothèse : options européennes sans dividendes

strike_actualise = K * np.exp(-r*T)

# Tolérance adaptée à l'échelle des prix
tolerance = 1e-8 * max(1.0, S0, strike_actualise)

# 1. Vérifier que les prix sont des nombres finis
assert np.isfinite(prix_put), "Le prix du put n'est pas fini."
assert np.isfinite(prix_call), "Le prix du call n'est pas fini."

# 2. Parité : call - put = S0 - K*exp(-r*T)
ecart_parite = prix_call - prix_put - (S0 - strike_actualise)
parite_ok = abs(ecart_parite) <= tolerance

# 3. Bornes du put
borne_inf_put = max(strike_actualise - S0, 0.0)
borne_sup_put = strike_actualise

put_ok = (
    borne_inf_put - tolerance
    <= prix_put
    <= borne_sup_put + tolerance
)

# 4. Bornes du call
borne_inf_call = max(S0 - strike_actualise, 0.0)
borne_sup_call = S0

call_ok = (
    borne_inf_call - tolerance
    <= prix_call
    <= borne_sup_call + tolerance
)

# 5. Affichage des résultats
print("VÉRIFICATION DE LA PARITÉ CALL-PUT")
print("-" * 65)
print(f"Call - Put         : {prix_call - prix_put:.10f}")
print(f"S0 - K*exp(-r*T)   : {S0 - strike_actualise:.10f}")
print(f"Écart de parité    : {ecart_parite:.3e}")
print("Résultat           :", "OK" if parite_ok else "ÉCHEC")

print()
print("VÉRIFICATION DES BORNES")
print("-" * 65)
print(
    f"Put  : {borne_inf_put:.10f} <= "
    f"{prix_put:.10f} <= {borne_sup_put:.10f}"
)
print("Résultat put       :", "OK" if put_ok else "ÉCHEC")

print(
    f"Call : {borne_inf_call:.10f} <= "
    f"{prix_call:.10f} <= {borne_sup_call:.10f}"
)
print("Résultat call      :", "OK" if call_ok else "ÉCHEC")

print()
print(f"Tolérance utilisée : {tolerance:.3e}")

# Signaler une éventuelle anomalie après avoir affiché les diagnostics
assert parite_ok and put_ok and call_ok, (
    "Au moins un contrôle a échoué : examiner les résultats ci-dessus."
)

print("Tous les contrôles passent.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. Fonction auxiliaire : recalcul complet pour un couple (N, L)
# ============================================================

def prix_put_convergence(nombre_termes, largeur):
    """
    Prix du put pour les paramètres actuels.
    Recalcule les bornes et tous les tableaux à chaque appel.
    """
    ks_conv = kappa + lambdA
    ts_conv = kappa * theta / ks_conv
    x_conv = np.log(S0 / K)

    bornes_conv = trunc_range(
        r, T, kappa, lambdA, theta,
        sig_0, sig_sig, rho,
        largeur, x_conv
    )

    a_conv, b_conv = bornes_conv[:2]

    if not a_conv < 0 < b_conv:
        raise ValueError(
            f"L={largeur} : l'intervalle ne vérifie pas a < 0 < b."
        )

    omega_conv = (
        np.arange(nombre_termes) * np.pi / (b_conv - a_conv)
    )

    phi_conv = carac_heston(
        omega_conv, r, T,
        ks_conv, ts_conv,
        sig_0, sig_sig, rho
    )

    chi_conv = chi(omega_conv, a_conv, a_conv, 0.0)
    psi_conv = psi(omega_conv, a_conv, a_conv, 0.0)

    U_conv = (
        2 / (b_conv - a_conv)
        * (psi_conv - chi_conv)
    )

    prix_conv = P(
        omega_conv, phi_conv, U_conv,
        x_conv, a_conv, K, r, T
    )

    if not np.isfinite(prix_conv):
        raise ValueError(
            f"Prix non fini pour N={nombre_termes}, L={largeur}."
        )

    return float(prix_conv)


# ============================================================
# 2. Paramètres de l'étude
# ============================================================

N_convergence = [16, 32, 64, 128, 256, 512, 1024]
L_convergence = [6, 8, 10, 12, 14, 16]

print("OPTION ÉTUDIÉE")
print(f"S0={S0}, K={K}, r={r}, T={T}")
print(f"rho={rho}, sig_0={sig_0}, sig_sig={sig_sig}")
print()


# ============================================================
# 3. Référence numérique et contrôle de sa stabilité
# ============================================================

# Cette référence est une approximation COS plus fine,
# pas une solution exacte ni une méthode indépendante.
reference_conv = prix_put_convergence(4096, 20)

# Vérification supplémentaire : plus de termes et intervalle élargi
reference_controle_conv = prix_put_convergence(8192, 24)

ecart_reference_conv = abs(
    reference_conv - reference_controle_conv
)

print(f"Référence COS (N=4096, L=20) : {reference_conv:.12f}")
print(f"Contrôle COS  (N=8192, L=24) : {reference_controle_conv:.12f}")
print(f"Écart entre références     : {ecart_reference_conv:.3e}")

seuil_reference_conv = 1e-8 * max(1.0, abs(reference_conv))

if ecart_reference_conv > seuil_reference_conv:
    print(
        "ATTENTION : la référence n'est pas suffisamment stable "
        "au seuil choisi. Les écarts ci-dessous sont à interpréter "
        "avec prudence."
    )

print()


# ============================================================
# 4. Calcul pour tous les couples (N, L)
# ============================================================

prix_convergence = np.empty(
    (len(L_convergence), len(N_convergence))
)

erreurs_convergence = np.empty_like(prix_convergence)

for i_conv, largeur_conv in enumerate(L_convergence):

    for j_conv, nombre_conv in enumerate(N_convergence):

        valeur_conv = prix_put_convergence(
            nombre_conv, largeur_conv
        )

        prix_convergence[i_conv, j_conv] = valeur_conv

        erreurs_convergence[i_conv, j_conv] = abs(
            valeur_conv - reference_conv
        )


# ============================================================
# 5. Tableau des prix et écarts
# ============================================================

print(
    f"{'L':>4} | {'N':>6} | {'Prix du put':>18} | "
    f"{'Écart à la référence':>22}"
)
print("-" * 62)

for i_conv, largeur_conv in enumerate(L_convergence):

    for j_conv, nombre_conv in enumerate(N_convergence):

        print(
            f"{largeur_conv:4d} | "
            f"{nombre_conv:6d} | "
            f"{prix_convergence[i_conv, j_conv]:18.10f} | "
            f"{erreurs_convergence[i_conv, j_conv]:22.3e}"
        )

    print("-" * 62)


# ============================================================
# 6. Graphiques de convergence
# ============================================================

fig_conv, axes_conv = plt.subplots(
    1, 2, figsize=(13, 4.5)
)

# Uniquement pour afficher les zéros sur une échelle logarithmique.
# Les erreurs originales restent inchangées dans le tableau.
plancher_affichage_conv = (
    1e-15 * max(1.0, abs(reference_conv))
)

# À gauche : influence de N pour plusieurs L
for i_conv, largeur_conv in enumerate(L_convergence):

    axes_conv[0].plot(
        N_convergence,
        np.maximum(
            erreurs_convergence[i_conv],
            plancher_affichage_conv
        ),
        "o-",
        label=f"L = {largeur_conv}"
    )

axes_conv[0].set_xscale("log", base=2)
axes_conv[0].set_yscale("log")
axes_conv[0].set_xlabel("Nombre de termes N")
axes_conv[0].set_ylabel("Écart absolu de prix à la référence")
axes_conv[0].set_title("Convergence selon N")
axes_conv[0].grid(True, which="both", alpha=0.3)
axes_conv[0].legend()

# À droite : influence de L pour plusieurs N
for nombre_conv in [64, 256, 1024]:

    j_conv = N_convergence.index(nombre_conv)

    axes_conv[1].plot(
        L_convergence,
        np.maximum(
            erreurs_convergence[:, j_conv],
            plancher_affichage_conv
        ),
        "o-",
        label=f"N = {nombre_conv}"
    )

axes_conv[1].set_yscale("log")
axes_conv[1].set_xlabel("Paramètre de largeur L")
axes_conv[1].set_ylabel("Écart absolu de prix à la référence")
axes_conv[1].set_title("Sensibilité à l'intervalle")
axes_conv[1].grid(True, which="both", alpha=0.3)
axes_conv[1].legend()

fig_conv.tight_layout()
plt.show()
```
